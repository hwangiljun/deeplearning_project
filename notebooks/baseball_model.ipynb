```python

import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader, WeightedRandomSampler
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import joblib
import os

try:
    if 'full_data' in locals():
        df = full_data.copy()
    else:
        df = pd.read_csv("mlb_pitch_data_2024_season.csv")
except:
    exit()

# 데이터 전처리
def define_outcome(row):
    evt = str(row['events'])
    desc = str(row['description'])
    if evt in ['strikeout', 'field_out', 'single', 'double', 'home_run', 'walk', 'hit_by_pitch']:
        return evt
    if 'ball' in desc or 'blocked' in desc: return 'ball'
    elif 'swinging_strike' in desc: return 'swinging_strike'
    elif 'foul' in desc: return 'foul'
    elif 'strike' in desc: return 'strike'
    return 'other'

df['outcome'] = df.apply(define_outcome, axis=1)
df = df[df['outcome'] != 'other'].reset_index(drop=True)

# 결측치 처리
req_cols = ['batter', 'pitcher', 'stand', 'p_throws', 'balls', 'strikes', 'outs_when_up', 'on_1b', 'on_2b', 'on_3b', 'pitch_type', 'release_speed', 'release_spin_rate', 'pfx_x', 'pfx_z', 'plate_x', 'plate_z', 'outcome']
for c in req_cols:
    if c not in df.columns: df[c] = 0
df = df[req_cols]

cols_to_zero = ['balls', 'strikes', 'outs_when_up', 'on_1b', 'on_2b', 'on_3b']
df[cols_to_zero] = df[cols_to_zero].fillna(0)

num_cols_to_fill = ['release_speed', 'release_spin_rate', 'pfx_x', 'pfx_z', 'plate_x', 'plate_z']
for col in num_cols_to_fill:
    df[col] = df[col].astype(float)
    df[col] = df[col].fillna(df[col].mean())

cat_cols_to_fill = ['pitch_type', 'stand', 'p_throws', 'pitcher', 'batter']
for col in cat_cols_to_fill:
    df[col] = df[col].fillna('Unknown')

df = df.dropna().reset_index(drop=True)

# 인코딩
encoders = {}
cat_cols = ['batter', 'pitcher', 'stand', 'p_throws', 'pitch_type', 'outcome']
for col in cat_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

for col in ['on_1b', 'on_2b', 'on_3b']:
    df[col] = df[col].apply(lambda x: 0 if pd.isna(x) or x == 0 else 1)

# 수치형 스케일링
num_cols = ['balls', 'strikes', 'outs_when_up', 'on_1b', 'on_2b', 'on_3b',
            'release_speed', 'release_spin_rate', 'pfx_x', 'pfx_z', 'plate_x', 'plate_z']
real_num_cols = [c for c in num_cols if c in df.columns]

scaler = MinMaxScaler()
df[real_num_cols] = scaler.fit_transform(df[real_num_cols])
encoders['scaler'] = scaler
encoders['num_cols'] = real_num_cols

#데이터 생성
SEQ_LEN = 8
sequences_cat = []
sequences_num = []
contexts = [] # 현재 타석의 상황 정보 (Skip Connection용)
targets = []

grouped = df.groupby('batter')
print(f"학습 데이터 생성 중")

# Context 컬럼: 볼카운트, 아웃, 주자 상황 (현재 시점의 중요 정보)
context_cols_idx = [df.columns.get_loc(c) for c in ['balls', 'strikes', 'outs_when_up', 'on_1b', 'on_2b', 'on_3b']]

# Feature 순서 정의
cat_feature_cols = ['batter', 'pitcher', 'pitch_type', 'stand', 'p_throws']
num_feature_cols = real_num_cols

for batter_id, group in tqdm(grouped):
    if len(group) > SEQ_LEN:
        data_cat = group[cat_feature_cols].values
        data_num = group[num_feature_cols].values
        # 전체 데이터프레임 값 (Context 추출용)
        data_all = group.values
        target_vals = group['outcome'].values

        stride = 2 if len(df) > 100000 else 1

        for i in range(0, len(data_cat) - SEQ_LEN, stride):
            # 1. 과거 시퀀스
            sequences_cat.append(data_cat[i : i+SEQ_LEN])
            sequences_num.append(data_num[i : i+SEQ_LEN])

            # 2. [현재 시점(마지막 공)의 Context 정보 추출 (볼, 스트라이크, 주자 등)
            # data_num의 마지막 행에서 앞쪽 6개 컬럼(볼,스,아웃,주자)이 Context라고 가정
            # (real_num_cols 순서: balls, strikes, outs, on_1b, on_2b, on_3b ...)
            current_context = data_num[i+SEQ_LEN-1][:6]
            contexts.append(current_context)

            # 3. 타겟
            targets.append(target_vals[i+SEQ_LEN - 1])

# 텐서 변환
X_cat = torch.tensor(np.array(sequences_cat), dtype=torch.long)
X_num = torch.tensor(np.array(sequences_num), dtype=torch.float32)
X_ctx = torch.tensor(np.array(contexts), dtype=torch.float32) # Context Tensor
y = torch.tensor(np.array(targets), dtype=torch.long)

# 데이터 분할
indices = np.arange(len(y))
train_idx, val_idx = train_test_split(indices, test_size=0.2, random_state=42, stratify=y)

# 샘플러
class_counts = np.bincount(y[train_idx].numpy())
class_weights = 1. / np.maximum(class_counts, 1)
sample_weights = class_weights[y[train_idx].numpy()]
sampler = WeightedRandomSampler(weights=sample_weights, num_samples=len(train_idx), replacement=True)

#  Focal Loss 정의 (어려운 샘플에 집중)
class FocalLoss(nn.Module):
    def __init__(self, alpha=1, gamma=2, reduction='mean'):
        super(FocalLoss, self).__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.reduction = reduction

    def forward(self, inputs, targets):
        CE_loss = F.cross_entropy(inputs, targets, reduction='none')
        pt = torch.exp(-CE_loss) # 예측 확률
        F_loss = self.alpha * (1-pt)**self.gamma * CE_loss

        if self.reduction == 'mean': return torch.mean(F_loss)
        elif self.reduction == 'sum': return torch.sum(F_loss)
        else: return F_loss

# Context-Aware Transformer (Skip Connection 추가)
class AdvancedBaseballTransformer(nn.Module):
    def __init__(self, embedding_dims, num_numeric, d_model, num_classes, nhead=8, num_layers=8):
        super().__init__()

        # Embeddings
        self.embeddings = nn.ModuleDict()
        total_embed_dim = 0
        for col, (num_unique, dim) in embedding_dims.items():
            self.embeddings[col] = nn.Embedding(num_unique, dim)
            total_embed_dim += dim

        self.input_proj = nn.Linear(total_embed_dim + num_numeric, d_model)
        self.pos_encoder = nn.Parameter(torch.zeros(1, SEQ_LEN, d_model))

        # Transformer
        encoder_layer = nn.TransformerEncoderLayer(d_model=d_model, nhead=nhead, batch_first=True, dropout=0.2, norm_first=True)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # Context(볼카운트 등)를 처리할 별도 레이어
        self.context_proj = nn.Linear(6, 32) # Context 차원(6) -> 32

        self.fc_head = nn.Sequential(
            nn.Linear(d_model + 32, 128), # 결합된 벡터 입력
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, num_classes)
        )

    def forward(self, x_cat, x_num, x_ctx):
        embedded_list = []
        keys = ['batter', 'pitcher', 'pitch_type', 'stand', 'p_throws']
        for i, key in enumerate(keys):
            embedded_list.append(self.embeddings[key](x_cat[:, :, i]))

        x_emb = torch.cat(embedded_list, dim=2)
        x_combined = torch.cat([x_emb, x_num], dim=2)

        x = self.input_proj(x_combined) + self.pos_encoder
        x = self.transformer(x)

        history_vec = x[:, -1, :]

        current_vec = F.relu(self.context_proj(x_ctx))

        final_vec = torch.cat([history_vec, current_vec], dim=1)

        return self.fc_head(final_vec)

# 학습 설정
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"연구 모델 학습 (Device: {device})")

embed_config = {
    'batter': (len(encoders['batter'].classes_), 32),
    'pitcher': (len(encoders['pitcher'].classes_), 32),
    'pitch_type': (len(encoders['pitch_type'].classes_), 8),
    'stand': (len(encoders['stand'].classes_), 2),
    'p_throws': (len(encoders['p_throws'].classes_), 2)
}
num_numeric_features = len(real_num_cols)
NUM_CLASSES = len(encoders['outcome'].classes_)

# 모델 생성 (d_model=256)
model = AdvancedBaseballTransformer(embed_config, num_numeric_features, d_model=256, num_classes=NUM_CLASSES, nhead=8, num_layers=8).to(device)

# [NEW] Focal Loss 사용 (Gamma=2로 설정하여 어려운 샘플 집중)
criterion = FocalLoss(gamma=2.0)
optimizer = optim.Adam(model.parameters(), lr=0.0005)
scheduler = optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.5, patience=2)

# DataLoader
train_ds = TensorDataset(X_cat[train_idx], X_num[train_idx], X_ctx[train_idx], y[train_idx])
val_ds = TensorDataset(X_cat[val_idx], X_num[val_idx], X_ctx[val_idx], y[val_idx])

train_loader = DataLoader(train_ds, batch_size=128, sampler=sampler)
val_loader = DataLoader(val_ds, batch_size=128, shuffle=False)

EPOCHS = 30
best_val_acc = 0.0

for epoch in range(EPOCHS):
    model.train()
    train_loss = 0
    for bx_cat, bx_num, bx_ctx, by in tqdm(train_loader, desc=f"Epoch {epoch+1}", leave=False):
        bx_cat, bx_num, bx_ctx, by = bx_cat.to(device), bx_num.to(device), bx_ctx.to(device), by.to(device)

        optimizer.zero_grad()
        out = model(bx_cat, bx_num, bx_ctx) # Context 추가 입력
        loss = criterion(out, by)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        train_loss += loss.item()

    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for bx_cat, bx_num, bx_ctx, by in val_loader:
            bx_cat, bx_num, bx_ctx, by = bx_cat.to(device), bx_num.to(device), bx_ctx.to(device), by.to(device)
            out = model(bx_cat, bx_num, bx_ctx)
            _, predicted = torch.max(out, 1)
            total += by.size(0)
            correct += (predicted == by).sum().item()

    val_acc = 100 * correct / total
    scheduler.step(val_acc)

    print(f"Epoch {epoch+1} | Loss: {train_loss/len(train_loader):.4f} | Val Acc: {val_acc:.2f}%", end="")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), 'paper_model.pth') # 이름은 그대로 유지 (호환성)
        print(" ->Model Saved!")
    else:
        print("")

print(f"\n최종 최고 정확도: {best_val_acc:.2f}%")

# 저장
joblib.dump(encoders, 'paper_encoders.pkl')
joblib.dump(embed_config, 'embed_config.pkl')
joblib.dump(real_num_cols, 'num_cols.pkl')

print("파일 다운로드")
from google.colab import files
try:
    files.download('paper_model.pth')
    files.download('paper_encoders.pkl')
    files.download('embed_config.pkl')
    files.download('num_cols.pkl')
except:
    pass
```

```python
# 비교군을 위한 모델(기존 연구 모델)
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader, WeightedRandomSampler
from sklearn.preprocessing import LabelEncoder, MinMaxScaler
from sklearn.model_selection import train_test_split
from tqdm import tqdm
import math
import os

# 1. 데이터 로드
try:
    if 'full_data' in locals():
        df = full_data.copy()
    else:
        df = pd.read_csv("mlb_pitch_data_2024_season.csv")
except:
    exit()

# 데이터 전처리
def define_outcome(row):
    evt = str(row['events'])
    desc = str(row['description'])
    if evt in ['strikeout', 'field_out', 'single', 'double', 'home_run', 'walk', 'hit_by_pitch']:
        return evt
    if 'ball' in desc or 'blocked' in desc: return 'ball'
    elif 'swinging_strike' in desc: return 'swinging_strike'
    elif 'foul' in desc: return 'foul'
    elif 'strike' in desc: return 'strike'
    return 'other'

df['outcome'] = df.apply(define_outcome, axis=1)
df = df[df['outcome'] != 'other'].reset_index(drop=True)

# 결측치 및 컬럼 처리
req_cols = ['batter', 'pitcher', 'stand', 'p_throws', 'balls', 'strikes', 'outs_when_up',
            'on_1b', 'on_2b', 'on_3b', 'pitch_type', 'release_speed', 'release_spin_rate',
            'pfx_x', 'pfx_z', 'plate_x', 'plate_z', 'outcome']

for c in req_cols:
    if c not in df.columns: df[c] = 0
df = df[req_cols]

cols_to_zero = ['balls', 'strikes', 'outs_when_up', 'on_1b', 'on_2b', 'on_3b']
df[cols_to_zero] = df[cols_to_zero].fillna(0)

num_cols_to_fill = ['release_speed', 'release_spin_rate', 'pfx_x', 'pfx_z', 'plate_x', 'plate_z']
for col in num_cols_to_fill:
    df[col] = df[col].astype(float)
    df[col] = df[col].fillna(df[col].mean())

cat_cols_to_fill = ['pitch_type', 'stand', 'p_throws', 'pitcher', 'batter']
for col in cat_cols_to_fill:
    df[col] = df[col].fillna('Unknown')

df = df.dropna().reset_index(drop=True)

# 인코딩
encoders = {}
cat_cols = ['batter', 'pitcher', 'stand', 'p_throws', 'pitch_type', 'outcome']
for col in cat_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col].astype(str))
    encoders[col] = le

for col in ['on_1b', 'on_2b', 'on_3b']:
    df[col] = df[col].apply(lambda x: 0 if pd.isna(x) or x == 0 else 1)

# 수치형 스케일링
num_cols = ['balls', 'strikes', 'outs_when_up', 'on_1b', 'on_2b', 'on_3b',
            'release_speed', 'release_spin_rate', 'pfx_x', 'pfx_z', 'plate_x', 'plate_z']
real_num_cols = [c for c in num_cols if c in df.columns]

scaler = MinMaxScaler()
df[real_num_cols] = scaler.fit_transform(df[real_num_cols])

# 데이터 생성

PAPER_SEQ_LEN = 20

sequences_cat = []
sequences_num = []
targets = []

grouped = df.groupby('batter')

cat_feature_cols = ['batter', 'pitcher', 'pitch_type', 'stand', 'p_throws']
num_feature_cols = real_num_cols

for batter_id, group in tqdm(grouped):
    if len(group) > PAPER_SEQ_LEN:
        data_cat = group[cat_feature_cols].values
        data_num = group[num_feature_cols].values
        target_vals = group['outcome'].values

        stride = 1

        for i in range(0, len(data_cat) - PAPER_SEQ_LEN, stride):
            seq_cat = data_cat[i : i+PAPER_SEQ_LEN].copy()
            seq_num = data_num[i : i+PAPER_SEQ_LEN].copy()

            sequences_cat.append(seq_cat)
            sequences_num.append(seq_num)

            targets.append(target_vals[i+PAPER_SEQ_LEN - 1])

# 텐서 변환
X_cat = torch.tensor(np.array(sequences_cat), dtype=torch.long)
X_num = torch.tensor(np.array(sequences_num), dtype=torch.float32)
y = torch.tensor(np.array(targets), dtype=torch.long)

# 데이터 분할
indices = np.arange(len(y))
train_idx, val_idx = train_test_split(indices, test_size=0.2, random_state=42, stratify=y)

# 샘플러
class_counts = np.bincount(y[train_idx].numpy())
class_weights = 1. / np.maximum(class_counts, 1)
sample_weights = class_weights[y[train_idx].numpy()]
sampler = WeightedRandomSampler(weights=sample_weights, num_samples=len(train_idx), replacement=True)

# DataLoader
train_ds = TensorDataset(X_cat[train_idx], X_num[train_idx], y[train_idx])
val_ds = TensorDataset(X_cat[val_idx], X_num[val_idx], y[val_idx])

train_loader = DataLoader(train_ds, batch_size=64, sampler=sampler)
val_loader = DataLoader(val_ds, batch_size=64, shuffle=False)

# Positional Encoding
class PositionalEncoding(nn.Module):
    def __init__(self, d_model, max_len=500):
        super(PositionalEncoding, self).__init__()
        pe = torch.zeros(max_len, d_model)
        position = torch.arange(0, max_len, dtype=torch.float).unsqueeze(1)
        div_term = torch.exp(torch.arange(0, d_model, 2).float() * (-math.log(10000.0) / d_model))
        pe[:, 0::2] = torch.sin(position * div_term)
        pe[:, 1::2] = torch.cos(position * div_term)
        pe = pe.unsqueeze(0)
        self.register_buffer('pe', pe)

    def forward(self, x):
        return x + self.pe[:, :x.size(1), :]

#  Baseline Transformer Architecture
class PaperBaselineTransformer(nn.Module):
    def __init__(self, embedding_dims, num_numeric, d_model=128, num_classes=10, nhead=4, num_layers=4):
        super().__init__()

        # 1. Embeddings
        self.embeddings = nn.ModuleDict()
        total_embed_dim = 0
        for col, (num_unique, dim) in embedding_dims.items():
            self.embeddings[col] = nn.Embedding(num_unique, dim)
            total_embed_dim += dim

        # Input Projection
        self.input_proj = nn.Linear(total_embed_dim + num_numeric, d_model)

        # 2. Positional Encoding
        self.pos_encoder = PositionalEncoding(d_model, max_len=PAPER_SEQ_LEN)

        # 3. Transformer Encoder
        encoder_layer = nn.TransformerEncoderLayer(d_model=d_model, nhead=nhead, batch_first=True, dropout=0.3)
        self.transformer = nn.TransformerEncoder(encoder_layer, num_layers=num_layers)

        # 4. Last Pitch Linear Transformation
        self.last_pitch_proj = nn.Linear(total_embed_dim + num_numeric, d_model)

        # 5. Output Heads
        self.fc_head = nn.Sequential(
            nn.Linear(d_model * 2, 128),
            nn.ReLU(),
            nn.Dropout(0.3), # FC Layer에도 Dropout 추가
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, num_classes)
        )

    def forward(self, x_cat, x_num):
        embedded_list = []
        keys = ['batter', 'pitcher', 'pitch_type', 'stand', 'p_throws']
        for i, key in enumerate(keys):
            embedded_list.append(self.embeddings[key](x_cat[:, :, i]))

        x_emb = torch.cat(embedded_list, dim=2)
        x_combined = torch.cat([x_emb, x_num], dim=2)

        x = self.input_proj(x_combined)
        x = self.pos_encoder(x)
        x_trans = self.transformer(x)

        context_vec = x_trans[:, -1, :]

        last_pitch_raw = x_combined[:, -1, :]
        last_pitch_vec = self.last_pitch_proj(last_pitch_raw)

        combined_vec = torch.cat([context_vec, last_pitch_vec], dim=1)

        return self.fc_head(combined_vec)


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"비교 모델 학습 시작 (Device: {device})")

embed_config = {
    'batter': (len(encoders['batter'].classes_), 32),
    'pitcher': (len(encoders['pitcher'].classes_), 32),
    'pitch_type': (len(encoders['pitch_type'].classes_), 8),
    'stand': (len(encoders['stand'].classes_), 2),
    'p_throws': (len(encoders['p_throws'].classes_), 2)
}
num_numeric_features = len(real_num_cols)
NUM_CLASSES = len(encoders['outcome'].classes_)

model = PaperBaselineTransformer(
    embed_config,
    num_numeric_features,
    d_model=128,
    num_classes=NUM_CLASSES,
    nhead=4,
    num_layers=4
).to(device)

criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.0001)

EPOCHS = 10
best_val_acc = 0.0

for epoch in range(EPOCHS):
    model.train()
    train_loss = 0
    for bx_cat, bx_num, by in tqdm(train_loader, desc=f"Baseline Epoch {epoch+1}", leave=False):
        bx_cat, bx_num, by = bx_cat.to(device), bx_num.to(device), by.to(device)

        optimizer.zero_grad()
        out = model(bx_cat, bx_num)
        loss = criterion(out, by)
        loss.backward()
        optimizer.step()
        train_loss += loss.item()

    model.eval()
    correct = 0
    total = 0
    with torch.no_grad():
        for bx_cat, bx_num, by in val_loader:
            bx_cat, bx_num, by = bx_cat.to(device), bx_num.to(device), by.to(device)
            out = model(bx_cat, bx_num)
            _, predicted = torch.max(out, 1)
            total += by.size(0)
            correct += (predicted == by).sum().item()

    val_acc = 100 * correct / total
    print(f"Epoch {epoch+1} | Loss: {train_loss/len(train_loader):.4f} | Val Acc: {val_acc:.2f}%", end="")

    if val_acc > best_val_acc:
        best_val_acc = val_acc
        torch.save(model.state_dict(), 'baseline_paper_model.pth')
        print(" -> Saved!")
    else:
        print("")

print(f"\n🏆 Baseline 최종 최고 정확도: {best_val_acc:.2f}%")
```
